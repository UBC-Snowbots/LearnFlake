```python
# Importing libraries

import gymnasium as gym
import robosuite as robs
import robosuite.wrappers.gym_wrapper as GymWrapper
from gym_wrapper import RobosuiteGymWrapper
from stable_baselines3 import PPO
```

```python
env = RobosuiteGymWrapper("Lift", robots="Panda", has_renderer=True, use_camera_obs=False, reward_shaping=True)

model = PPO.load("ppo_lift_model", env = env) # Loads trained model

model = PPO("MlpPolicy", env, verbose=1)

# Making the environment, loading trained model and declaring policy
```

```python
# Train the model
model.learn(total_timesteps=1000)
```

```python
from stable_baselines3.common.evaluation import evaluate_policy
mean_reward, std_reward = evaluate_policy(model, env, n_eval_episodes=10, render=True)

# Testing stage

model.save("ppo_lift_model")
```
